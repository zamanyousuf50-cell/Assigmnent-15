```python
pd.read_sql_query("""
SELECT name, area, population
FROM facts
WHERE area > 5000
""", conn)
```

```python
pd.read_sql_query("""
SELECT cities.name AS capital_name
FROM cities
INNER JOIN facts
ON cities.facts_id = facts.id
WHERE cities.capital = 1
AND facts.death_rate > 2
AND facts.birth_rate < 20
""", conn)
```

```python
pd.read_sql_query("""
SELECT cities.name AS city_name,
       facts.population_growth
FROM cities
INNER JOIN facts
ON cities.facts_id = facts.id
WHERE facts.population_growth BETWEEN 1 AND 5
ORDER BY facts.population_growth DESC
""", conn)
```

```python
pd.read_sql_query("""
SELECT facts.name AS country_name,
       cities.name AS city_name
FROM facts
LEFT JOIN cities
ON facts.id = cities.facts_id
WHERE facts.name LIKE 'A%'
AND cities.id IS NULL
""", conn)
```

```python
pd.read_sql_query("""
SELECT cities.name AS capital_name,
       facts.death_rate
FROM cities
LEFT JOIN facts
ON cities.facts_id = facts.id
WHERE cities.capital = 1
""", conn)
```
